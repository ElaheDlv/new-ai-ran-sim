```python
import itertools

PARAMS = {
    "model": ["lstm", "tcn", "seq2seq"],
    "window": [128, 256, 384],
    "batch_size": [32, 64],
    "hidden_dim": [128, 192, 256],
    "num_layers": [2, 3, 4, 8],
    "dropout": [0.2, 0.3, 0.4],
    "optimizer": ["adam", "adamw", "rmsprop"],
    "learning_rate": ["1e-3","3e-4", "5e-4"],
    "loss": ["mse", "smoothl1", "mae", "huber", "l1"],
    "feature_sets": ["time+delta_t+length"],
    "lr_scheduler": ["none","plateau", "cosine"],
    #"early_stop": [30],
}

with open("config.txt", "w") as f:
    keys, values = zip(*PARAMS.items())
    for combo in itertools.product(*values):
        config = dict(zip(keys, combo))
        line = " ".join([f"--{k.replace('_','-')} {v}" for k,v in config.items()])
        f.write(line + "\n")
```
